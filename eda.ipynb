```python
import matplotlib.pyplot as plt

stats.to_excel("../output3/stats.xlsx")
img_columns = ["mean", "std", "median"]
max_rows = 40
subset = stats[img_columns].head(max_rows)

fig, ax = plt.subplots(figsize=(len(img_columns) * 3, max_rows * 0.4 + 2))
ax.axis("off")
table = ax.table(
            cellText=subset.values,
            colLabels=subset.columns,
            rowLabels=subset.index,
            loc="center",
            cellLoc="center"
        )
table.auto_set_font_size(False)
table.set_fontsize(9)
table.scale(1.2, 1.2)
plt.title("Summary Stats (subset)", fontsize=14, weight="bold")
plt.savefig("../output3/stats.png", bbox_inches="tight")
plt.close(fig)
```
